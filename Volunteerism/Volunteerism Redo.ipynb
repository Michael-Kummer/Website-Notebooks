```python
fullData = cleanData.copy()

# Group questions by what's after the colon
volunteering_groups = {}

# First, group columns by their question text
for col in volunteeringBehaviour:
    if ':' in col:
        option = col.split(':', 1)[0].strip()
        question_text = col.split(':', 1)[1].strip()
        
        # Remove any conditional text in square brackets for the question name
        if '[' in question_text:
            question_text = question_text.split('[', 1)[0].strip()
            
        if question_text not in volunteering_groups:
            volunteering_groups[question_text] = []
        
        volunteering_groups[question_text].append((col, option))
    else:
        # Questions without colons get copied directly
        new_df[col] = renamedOnline22[col]

# Now process each question group
for question, columns in volunteering_groups.items():
    # For binary columns (Selected/Not selected pattern)
    if "Which, if any, type(s) of volunteering" in question:
        selected_options = []
        
        for idx in renamedOnline22.index:
            row_options = []
            for col, option in columns:
                # Check if this option was selected
                if pd.notna(renamedOnline22.loc[idx, col]) and "Selected" in str(renamedOnline22.loc[idx, col]):
                    row_options.append(option)
            
            selected_options.append(', '.join(row_options) if row_options else np.nan)
        
        new_df[question] = selected_options
    
    # For "Which areas did you formally volunteer in" question
    elif "Which areas did you formally volunteer in" in question:
        selected_areas = []
        
        for idx in renamedOnline22.index:
            areas = []
            for col, option in columns:
                if pd.notna(renamedOnline22.loc[idx, col]) and "Selected" in str(renamedOnline22.loc[idx, col]):
                    areas.append(option)
            
            selected_areas.append(', '.join(areas) if areas else np.nan)
        
        new_df[question] = selected_areas
    
    # For "Was any of the formal volunteer work" question
    elif "Was any of the formal volunteer work" in question:
        selected_types = []
        
        for idx in renamedOnline22.index:
            types = []
            for col, option in columns:
                if pd.notna(renamedOnline22.loc[idx, col]) and "Selected" in str(renamedOnline22.loc[idx, col]):
                    types.append(option)
            
            selected_types.append(', '.join(types) if types else np.nan)
        
        new_df[question] = selected_types
    
    # For "How did you find out about" question
    elif "How did you find out about" in question:
        selected_sources = []
        
        for idx in renamedOnline22.index:
            sources = []
            for col, option in columns:
                if pd.notna(renamedOnline22.loc[idx, col]) and "Selected" in str(renamedOnline22.loc[idx, col]):
                    sources.append(option)
            
            selected_sources.append(', '.join(sources) if sources else np.nan)
        
        new_df[question] = selected_sources
```
