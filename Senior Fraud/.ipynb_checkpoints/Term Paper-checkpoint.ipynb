```python
rawdata = pd.read_csv('FraudData.csv')
rawdata = rawdata.drop([
    "Type de plainte reçue",
    "Catégories thématiques sur la fraude et la cybercriminalité",
    "Genre",
    "Méthode de sollicitation",
    "Type de plainte",
    "Langue de correspondance",
    "Province/État",
    "Pays"
], axis=1)

fraudData = rawdata.rename(columns={"Numéro d'identification / Number ID": "Number ID", 
                                    "Date Received / Date reçue":"Date Received",
                                    "Victim Age Range / Tranche d'âge des victimes":  "Age",
                                    "Number of Victims / Nombre de victimes":"Number of Victims",
                                    "Dollar Loss /pertes financières":"Dollar Loss",
                                    "Fraud and Cybercrime Thematic Categories":"Category",
                                    "Language of Correspondence":"Language",
                                    "Solicitation Method":"SolicitMethod"})

#Clean Values
# Remove leading apostrophes
fraudData.loc[:, "Age"] = fraudData["Age"].str.replace("^'", "", regex=True)
fraudData.loc[fraudData["Age"] == "Not Available / non disponible", "Age"] = pd.NA
fraudData.loc[fraudData["Age"] == "Deceased / Décédé", "Age"] = pd.NA
fraudData.loc[fraudData["Age"] == "Business / Entreprise", "Age"] = pd.NA
fraudData.loc[fraudData["Gender"] == "Not Available", "Gender"] = pd.NA
fraudData.loc[fraudData["Gender"] == "Other", "Gender"] = pd.NA
fraudData.loc[fraudData["Gender"] == "Unknown", "Gender"] = pd.NA
fraudData.loc[fraudData["Gender"] == "Prefer not to say", "Gender"] = pd.NA
fraudData.loc[fraudData["Language"] == "Not Available", "Language"] = pd.NA
fraudData.loc[fraudData["Province/State"] == "Not Available", "Province/State"] = pd.NA
fraudData.loc[fraudData["SolicitMethod"] == "Other/unknown", "SolicitMethod"] = pd.NA
fraudData.loc[fraudData["SolicitMethod"] == "Not Available", "SolicitMethod"] = pd.NA
fraudData.loc[fraudData["Category"] == "Unknown", "Category"] = pd.NA
fraudData.loc[fraudData["Category"] == "Incomplete", "Category"] = pd.NA

# Have to remove the $ and comma to get an actual integer
fraudData['Dollar Loss'] = (fraudData['Dollar Loss']
                            .str.replace('$', '', regex=False)
                            .str.replace(',', '', regex=False)
                            .astype(float)
                            .astype(int))
#Convert Date/Time to Pandas + sort
fraudData["Date Received"] = pd.to_datetime(fraudData["Date Received"])
fraudData = fraudData.sort_values("Date Received")
fraudData['Year'] = pd.DatetimeIndex(fraudData['Date Received']).year
#
fraudData.loc[fraudData["Complaint Type"] == "Attempt", "Dollar Loss"] = pd.NA


#Make two new Dataframes - One that is for individuals who were
#scammed and one for those who put in a complaint
victimData  = fraudData[fraudData["Complaint Type"]== "Victim"]
attemptData = fraudData[fraudData["Complaint Type"]== "Attempt"]

fraudData.head()
```

```python


nonZeroVictim = victimData[victimData["Dollar Loss"]>0]

analysisData = nonZeroVictim.loc[
    lambda x: (x["Country"].notnull())
    & (x["Language"].notnull())
    & (x["Category"].notnull())
    & (x["Age"].notnull())
    & (x["Year"].notnull())
].dropna()


analysisData["lnDollarLoss"] = np.log(analysisData["Dollar Loss"])
analysisData["Senior"] = analysisData["Age"].isin(seniorAgeList)
analysisData = analysisData[analysisData["Gender"].isin(["Male", "Female"])]
analysisData = analysisData.rename(columns = {"Province/State":"Location"})

#Redefining and adding Vars
analysisData["Age"] = analysisData["Age"].replace({"1 - 9":"1 - 20", "10 - 19":"1 - 20"})
analysisData["Age"] = analysisData["Age"].replace({"90 - 99":"90+", "100 +":"90+"})
analysisData["Category"] = analysisData["Category"].replace({"GRANT":"Grant"})
analysisData["SolicitMethod"] = analysisData["SolicitMethod"].replace({"Internet-social network":"Social Media"})
analysisData["Category"] = analysisData["Category"].replace({"Emergency (Jail, Accident, Hospital, Help)":"Emergency"})


analysisData["EffortLevel"] = analysisData["Category"].apply(lambda x: "High Effort" if x in highEffortList else "Low Effort")

#Reindexing Year
analysisData['Year'] = analysisData['Year'] - 2021 #Start at 0

analysisData.head()
```

```python
import statsmodels.api as sm
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from statsmodels.stats.diagnostic import het_breuschpagan
model = smf.ols('lnDollarLoss ~ C(Category, Treatment(reference="Emergency")) + Gender + Senior + Year +C(SolicitMethod, Treatment(reference="Direct call")) + C(Category, Treatment(reference="Emergency")):Senior +C(Category, Treatment(reference="Emergency")):Gender', data=robustnessAnalysis).fit()
# After fitting your OLS model (assuming 'model' is your fitted statsmodels model)
# Run Breusch-Pagan test
bp_test = het_breuschpagan(model.resid, model.model.exog)
print(f"Breusch-Pagan p-value: {bp_test[1]}")  # p-value < 0.05 suggests heteroskedasticity

# Visual diagnostics
plt.figure(figsize=(10, 6))
plt.scatter(model.fittedvalues, model.resid)
plt.axhline(y=0, color='r', linestyle='-')
plt.xlabel('Fitted values')
plt.ylabel('Residuals')
plt.title('Residuals vs Fitted Values')
plt.show()

# Boxplots of residuals by category
plt.figure(figsize=(10, 6))
sns.boxplot(x='Senior', y=model.resid, data=analysisData)
plt.title('Residuals by Senior Status')
plt.show()

# If heteroskedasticity is detected, use robust standard errors
robust_cov = sm.robust.robust_linear.RLM(model.model.endog, model.model.exog).fit().bcov_scaled
robust_se = np.sqrt(np.diag(robust_cov))
print("Robust Standard Errors:")
print(pd.DataFrame({'Coefficient': model.params, 
                    'Original SE': model.bse,
                    'Robust SE': robust_se}))
```

# Multicolinearity

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from statsmodels.stats.diagnostic import het_breuschpagan

# Assuming your model is already fit:
model = smf.ols('lnDollarLoss ~ C(Category, Treatment(reference="Emergency")) + Gender + Senior + Year +C(SolicitMethod, Treatment(reference="Direct call"))', data=robustnessAnalysis).fit()

# Run Breusch-Pagan test
bp_test = het_breuschpagan(model.resid, model.model.exog)
print(f"Breusch-Pagan p-value: {bp_test[1]}")  # p-value < 0.05 suggests heteroskedasticity

# Visual diagnostics
plt.figure(figsize=(10, 6))
plt.scatter(model.fittedvalues, model.resid)
plt.axhline(y=0, color='r', linestyle='-')
plt.xlabel('Fitted values')
plt.ylabel('Residuals')
plt.title('Residuals vs Fitted Values')
plt.show()

# Boxplots of residuals by category
plt.figure(figsize=(10, 6))
sns.boxplot(x='Senior', y=model.resid, data=analysisData)
plt.title('Residuals by Senior Status')
plt.show()

# FIXED: Correct way to get robust standard errors
robust_results = model.get_robustcov_results(cov_type='HC3')
print("\nRobust Standard Errors (HC3):")
print(pd.DataFrame({
    'Coefficient': robust_results.params,
    'Original SE': model.bse,
    'Robust SE': robust_results.bse
}))

# For a quick summary comparison
print("\nSignificance comparison:")
print(pd.DataFrame({
    'Original p-value': model.pvalues,
    'Robust p-value': robust_results.pvalues
}))
```

```python
# Format the output with better precision
pd.set_option('display.precision', 4)  # Set display precision

# Create a more readable comparison DataFrame
comparison_df = pd.DataFrame({
    'Coefficient': model.params,
    'Original SE': model.bse,
    'Robust SE': robust_results.bse,
    'Original p-value': model.pvalues,
    'Robust p-value': robust_results.pvalues
})

# Display only the rows where significance changes
significance_changes = comparison_df[
    ((model.pvalues < 0.05) & (robust_results.pvalues >= 0.05)) |
    ((model.pvalues >= 0.05) & (robust_results.pvalues < 0.05))
]

print("Coefficients where significance changes after robust SE adjustment:")
print(significance_changes)

# Display the ratio of robust to original SE to see the most affected variables
comparison_df['SE_Ratio'] = robust_results.bse / model.bse
print("\nVariables most affected by heteroskedasticity (largest SE ratio changes):")
print(comparison_df.sort_values('SE_Ratio', ascending=False).head(10)[['Coefficient', 'Original SE', 'Robust SE', 'SE_Ratio']])
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
import matplotlib.pyplot as plt
import seaborn as sns
from statsmodels.stats.outliers_influence import variance_inflation_factor
from patsy import dmatrices

# Assuming your dataframe is named 'robustnessAnalysis'
# First, let's check VIFs

# Create a function to calculate VIF for each variable
def calculate_vif(df, formula):
    # Extract the variables from the formula (this uses patsy)
    y, X = dmatrices(formula, df, return_type='dataframe')
    
    # Calculate VIF for each predictor
    vif_data = pd.DataFrame()
    vif_data["Variable"] = X.columns
    vif_data["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]
    
    return vif_data.sort_values("VIF", ascending=False)

# Run the VIF calculation on your main model formula
vif_results = calculate_vif(robustnessAnalysis, 
                            'lnDollarLoss ~ C(Category, Treatment(reference="Emergency")) + Gender + Senior + Year + C(SolicitMethod, Treatment(reference="Direct call")) + C(Category, Treatment(reference="Emergency")):Senior + C(Category, Treatment(reference="Emergency")):Gender')

print("Variance Inflation Factors (VIF > 10 indicates problematic multicollinearity):")
print(vif_results)

# Create a correlation matrix for numeric variables
numeric_vars = robustnessAnalysis.select_dtypes(include=[np.number])
correlation_matrix = numeric_vars.corr()

# Visualize the correlation matrix
plt.figure(figsize=(12, 10))
heatmap = sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', linewidths=0.5, vmin=-1, vmax=1)
plt.title('Correlation Matrix of Numeric Variables')
plt.tight_layout()
plt.show()

# Check for multicollinearity in categorical variables using Cramer's V
def cramers_v(x, y):
    confusion_matrix = pd.crosstab(x, y)
    chi2 = sm.stats.Table(confusion_matrix).test_nominal_association().statistic
    n = confusion_matrix.sum().sum()
    phi2 = chi2/n
    r, k = confusion_matrix.shape
    phi2corr = max(0, phi2-((k-1)*(r-1))/(n-1))
    rcorr = r-((r-1)**2)/(n-1)
    kcorr = k-((k-1)**2)/(n-1)
    return np.sqrt(phi2corr/min((rcorr-1), (kcorr-1)))

# Apply Cramer's V to categorical pairs
categorical_vars = robustnessAnalysis.select_dtypes(include=['object', 'category'])
cat_pairs = []
for i, col1 in enumerate(categorical_vars.columns):
    for col2 in categorical_vars.columns[i+1:]:
        cat_pairs.append((col1, col2, cramers_v(categorical_vars[col1], categorical_vars[col2])))

# Create DataFrame of categorical correlations
cat_correlations = pd.DataFrame(cat_pairs, columns=['Variable 1', 'Variable 2', "Cramer's V"])
cat_correlations = cat_correlations.sort_values("Cramer's V", ascending=False)
print("\nHighest categorical variable associations (Cramer's V):")
print(cat_correlations.head(10))

# Check condition number (eigenvalues of correlation matrix)
# Values > 30 indicate potential issues, > 100 indicate severe issues
X = sm.add_constant(numeric_vars)
eigenvalues = np.linalg.eigvals(X.T.dot(X))
condition_number = np.sqrt(np.max(eigenvalues) / np.min(eigenvalues))
print(f"\nCondition Number: {condition_number:.2f}")
if condition_number < 30:
    print("Condition number is acceptable (< 30)")
elif condition_number < 100:
    print("Condition number indicates potential multicollinearity issues (30-100)")
else:
    print("Condition number indicates severe multicollinearity (> 100)")

# Auxiliary regressions for key variables
# This helps identify which variables are collinear with each other
key_variables = ['Senior', 'Gender', 'Year']
for var in key_variables:
    formula = f"{var} ~ " + " + ".join([v for v in key_variables if v != var]) + " + C(Category) + C(SolicitMethod)"
    model = smf.ols(formula, data=robustnessAnalysis).fit()
    print(f"\nR-squared for {var} predicted by other variables: {model.rsquared:.4f}")
    if model.rsquared > 0.8:
        print(f"WARNING: {var} shows high collinearity (R² > 0.8)")
```
