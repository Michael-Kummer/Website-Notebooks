# Data Cleaning

```python
rawdata = pd.read_csv('FraudData.csv')
rawdata = rawdata.drop(["Type de plainte recue",
                        "Categories thematiques sur la fraude et la cybercriminalite",
                        "Genre",
                        "Methode de sollicitation",
                        "Type de plainte",
                        "Langue de correspondance",
                        "Province/Etat",
                        "Pays"],axis=1)

fraudData = rawdata.rename(columns={"Numero d'identification / Number ID": "Number ID", 
                                    "Date Received / Date recue":"Date Received",
                                    "Victim Age Range / Tranche d'age des victimes":  "Age",
                                    "Number of Victims / Nombre de victimes":"Number of Victims",
                                    "Dollar Loss /pertes financieres":"Dollar Loss",
                                    "Fraud and Cybercrime Thematic Categories":"Category",
                                    "Language of Correspondence":"Language",
                                    "Solicitation Method":"SolicitMethod"})

#Clean Values
# Remove leading apostrophes
fraudData.loc[:, "Age"] = fraudData["Age"].str.replace("^'", "", regex=True)
fraudData.loc[fraudData["Age"] == "Not Available / non disponible", "Age"] = pd.NA
fraudData.loc[fraudData["Age"] == "Deceased / Décédé", "Age"] = pd.NA
fraudData.loc[fraudData["Age"] == "Business / Entreprise", "Age"] = pd.NA
fraudData.loc[fraudData["Gender"] == "Not Available", "Gender"] = pd.NA
fraudData.loc[fraudData["Gender"] == "Other", "Gender"] = pd.NA
fraudData.loc[fraudData["Gender"] == "Unknown", "Gender"] = pd.NA
fraudData.loc[fraudData["Gender"] == "Prefer not to say", "Gender"] = pd.NA
fraudData.loc[fraudData["Language"] == "Not Available", "Language"] = pd.NA
fraudData.loc[fraudData["Province/State"] == "Not Available", "Province/State"] = pd.NA
fraudData.loc[fraudData["SolicitMethod"] == "Other/unknown", "SolicitMethod"] = pd.NA
fraudData.loc[fraudData["SolicitMethod"] == "Not Available", "SolicitMethod"] = pd.NA
fraudData.loc[fraudData["Category"] == "Unknown", "Category"] = pd.NA

# Have to remove the $ and comma to get an actual integer
fraudData['Dollar Loss'] = (fraudData['Dollar Loss']
                            .str.replace('$', '', regex=False)
                            .str.replace(',', '', regex=False)
                            .astype(float)
                            .astype(int))
#Convert Date/Time to Pandas + sort
fraudData["Date Received"] = pd.to_datetime(fraudData["Date Received"])
fraudData = fraudData.sort_values("Date Received")
fraudData['Year'] = pd.DatetimeIndex(fraudData['Date Received']).year
#
fraudData.loc[fraudData["Complaint Type"] == "Attempt", "Dollar Loss"] = pd.NA


#Make two new Dataframes - One that is for individuals who were
#scammed and one for those who put in a complaint
victimData  = fraudData[fraudData["Complaint Type"]== "Victim"]
attemptData = fraudData[fraudData["Complaint Type"]== "Attempt"]

fraudData.head()
```

# Actual Analysis

```python
analysisData = nonZeroVictim.loc[
    lambda x: (x["Country"].notnull())
    & (x["Language"].notnull())
    & (x["Category"].notnull())
    & (x["Age"].notnull())
    & (x["Year"].notnull())
].dropna()

analysisData["lnDollarLoss"] = np.log(analysisData["Dollar Loss"])
analysisData["Senior"] = analysisData["Age"].isin(seniorAgeList)
analysisData = analysisData[analysisData["Gender"].isin(["Male", "Female"])]
analysisData = analysisData.rename(columns = {"Province/State":"Location"})

# Center the Year variable 
analysisData['Year_Centered'] = analysisData['Year'] - analysisData['Year'].mean()

#non ln() 
nonLnAnalysisData = analysisData.rename(columns = {"Dollar Loss":"DollarLoss"})

analysisData.head()
```

```python
analysisData["SolicitMethod"].unique()
```

```python
#Combination Data

highEffortList = [
    "Romance", 
    "Investments", 
    "Foreign Money Offer", 
    "Timeshare", 
    "Recovery Pitch", 
    "Job", 
    "Bank Investigator", 
    "GRANT", 
    "Pyramid"
]

lowEffortList = [
    "Vendor Fraud", 
    "Merchandise", 
    "Extortion", 
    "Emergency (Jail, Accident, Hospital, Help)", 
    "Spear Phishing", 
    "Counterfeit Merchandise", 
    "Service", 
    "Loan", 
    "Prize", 
    "False Billing", 
    "Health", 
    "Charity / Donation", 
    "Vacation", 
    "Unauthorized Charge", 
    "Survey", 
    "Collection Agency", 
    "Spoofing", 
    "Directory", 
    "Credit Card", 
    "Telecom Fraud", 
    "Psychics", 
]

comboData = analysisData
comboData["EffortLevel"] = comboData["Category"].apply(lambda x: "High Effort" if x in highEffortList else "Low Effort")
```

```python
# Run your regressions first
reg1 = smf.ols(
    'lnDollarLoss ~ Senior + Gender + C(EffortLevel, Treatment(reference="Low Effort")) + Year_Centered', 
    data=comboData
).fit()

reg2 = smf.ols(
    'lnDollarLoss ~ Senior + Gender + C(EffortLevel, Treatment(reference="Low Effort")) + Year_Centered + SolicitMethod', 
    data=comboData
).fit()

reg3 = smf.ols(
    'lnDollarLoss ~ Senior + Gender + C(EffortLevel, Treatment(reference="Low Effort")) + Year_Centered + SolicitMethod + Senior:C(EffortLevel, Treatment(reference="Low Effort")) + Gender:C(EffortLevel, Treatment(reference="Low Effort")) + Senior:Gender:C(EffortLevel, Treatment(reference="Low Effort"))', 
    data=comboData
).fit()

# Create Stargazer object
stargazer = Stargazer([reg1, reg2, reg3])

# First set the order using ORIGINAL names
stargazer.covariate_order([
    'Intercept',
    'C(EffortLevel, Treatment(reference="Low Effort"))[T.High Effort]',
    'Senior[T.True]',
    'Gender[T.Male]',
    'Senior[T.True]:C(EffortLevel, Treatment(reference="Low Effort"))[T.High Effort]',
    'Gender[T.Male]:C(EffortLevel, Treatment(reference="Low Effort"))[T.High Effort]',
    'Year_Centered',
    'SolicitMethod[T.Door to door/in person]',
    'SolicitMethod[T.Email]',
    'SolicitMethod[T.Internet-social network]',
    'SolicitMethod[T.Internet]',
    'SolicitMethod[T.Mail]',
    'SolicitMethod[T.Print]',
    'SolicitMethod[T.Radio]',
    'SolicitMethod[T.Television]',
    'SolicitMethod[T.Text message]',
    'SolicitMethod[T.Video Call]'
])

# THEN apply renaming
stargazer.rename_covariates({
    'C(EffortLevel, Treatment(reference="Low Effort"))[T.High Effort]': 'High Effort',
    'Senior[T.True]:C(EffortLevel, Treatment(reference="Low Effort"))[T.High Effort]': 'Senior × High Effort',
    'Gender[T.Male]:C(EffortLevel, Treatment(reference="Low Effort"))[T.High Effort]': 'Male × High Effort',
    'Senior[T.True]': 'Senior',
    'Gender[T.Male]': 'Male',
    'Year_Centered': 'Year (centered)',
    'SolicitMethod[T.Door to door/in person]': 'Solicit: Door to door',
    'SolicitMethod[T.Email]': 'Solicit: Email', 
    'SolicitMethod[T.Internet-social network]': 'Solicit: Social network',
    'SolicitMethod[T.Internet]': 'Solicit: Internet',
    'SolicitMethod[T.Mail]': 'Solicit: Mail',
    'SolicitMethod[T.Print]': 'Solicit: Print',
    'SolicitMethod[T.Radio]': 'Solicit: Radio',
    'SolicitMethod[T.Television]': 'Solicit: Television',
    'SolicitMethod[T.Text message]': 'Solicit: Text message',
    'SolicitMethod[T.Video Call]': 'Solicit: Video Call'
})

# Add custom column labels
stargazer.custom_columns(['Model 1', 'Model 2', 'Interaction Model'])

# Display the table
stargazer
```

```python
# Run your regressions first
reg1 = smf.ols(
    "lnDollarLoss ~ Senior + Gender + C(EffortLevel, Treatment(reference='Low Effort')) + Year_Centered", 
    data=comboData
).fit()

reg2 = smf.ols(
    "lnDollarLoss ~ Senior + Gender + C(EffortLevel, Treatment(reference='Low Effort')) + Year_Centered + SolicitMethod", 
    data=comboData
).fit()

# Fixed reg3 - removed extra closing parenthesis and problematic interaction term
reg3 = smf.ols(
    "lnDollarLoss ~ Senior + Gender + C(EffortLevel, Treatment(reference='Low Effort')) + Year_Centered + SolicitMethod + Senior:C(EffortLevel, Treatment(reference='Low Effort')) + Gender:C(EffortLevel, Treatment(reference='Low Effort')) + Senior:Gender:C(EffortLevel, Treatment(reference='Low Effort')) + SolicitMethod:C(EffortLevel, Treatment(reference='Low Effort'))", 
    data=comboData
).fit()

# FIRST check the actual parameter names
print("Actual parameter names in Model 3:")
for name in reg3.params.index:
    print(f"'{name}',")

# Create Stargazer object
stargazer = Stargazer([reg1, reg2, reg3])

# Now use the exact parameter names from the output above for:
# 1. covariate_order 
# 2. rename_covariates dictionary

# Add custom column labels
stargazer.custom_columns(['Demographics Only', 'With Methods', 'Interaction Model'])

# Display the table
stargazer
```

```python
stargazer.covariate_order(keep_covs, restrict=True)

stargazer.rename_covariates({
    'C(EffortLevel, Treatment(reference="Low Effort"))[T.High Effort]': 'High Effort',
    'Senior[T.True]:C(EffortLevel, Treatment(reference="Low Effort"))[T.High Effort]': 'Senior × High Effort',
    'Gender[T.Male]:C(EffortLevel, Treatment(reference="Low Effort"))[T.High Effort]': 'Male × High Effort',
    'Senior[T.True]': 'Senior',
    'Gender[T.Male]': 'Male',
    'Year_Centered': 'Year (centered)',
})

stargazer.custom_columns(['Demographics Only', 'With Methods', 'Methods & Interaction'])
stargazer.title("Donation Amount Analysis (Significant Methods Only)")
stargazer
```

```python
import matplotlib.pyplot as plt

# Create a copy of the data
agePieData = analysisData.copy()

# IMPORTANT: Use replace instead of rename
agePieData['Age'] = agePieData['Age'].replace({"90 - 99":"90+", "100 +":"90+"})

# Define age order for better visualization
age_order = ['1 - 9', '10 - 19', '20 - 29', '30 - 39', '40 - 49', 
             '50 - 59', '60 - 69', '70 - 79', '80 - 89', '90+']

# Get counts and sort by our defined order
counts = agePieData['Age'].value_counts()
counts = counts.reindex(age_order).dropna()

# Remove very small categories (optional)
min_threshold = 0.01 * counts.sum()  # 1% threshold
counts = counts[counts >= min_threshold]

# Create figure with transparent background
fig = plt.figure(figsize=(8, 6), facecolor='none')  
plt.pie(
    counts, 
    labels=counts.index, 
    autopct='%1.1f%%',
    startangle=90,
    pctdistance=0.85,  # Move percentage labels closer to center
    labeldistance=1.1,  # Move labels further out
)

plt.title('Age Distribution of Fraud Victims', fontsize=16)
plt.axis('equal')
plt.savefig('age_distribution.png', transparent=True, bbox_inches='tight')
plt.show()
```
